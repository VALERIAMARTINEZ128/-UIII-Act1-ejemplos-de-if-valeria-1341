#### ejemplo de if


```python
# Ejemplo 1: Sentencia if simple con Operador Relacional
# Se utiliza para ejecutar un bloque de código solo si se cumple una condición específica

edad = 20
if edad >= 18:
    print("Eres mayor de edad.")
```

```python
# Ejemplo 1: Sentencia if-else
# Se utiliza para decidir entre dos caminos posibles

edad = 15  # Cambiamos el valor para probar la condición
if edad >= 18:
    print("Eres mayor de edad.")
else:
    print("Eres menor de edad.")
```

```python
numero = 15
if numero % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")
```

```python
numero = 15

# Primero verificamos si es par o impar
if numero % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")

# Ahora verificamos si es positivo, negativo o cero
if numero > 0:
    print("El número es positivo.")
elif numero < 0:
    print("El número es negativo.")
else:
    print("El número es cero.")
```

```python
# Ejemplo 3: Sentencia if-elif-else con Operadores Lógicos y Relacionales
# Se utiliza cuando existen múltiples condiciones mutuamente excluyentes
puntaje = 85
if puntaje >= 90:
    print("Calificación: A")
elif puntaje >= 80:
    print("Calificación: B")
else:
    print("Calificación: C o inferior")
puntaje = 85
if puntaje >= 90:
    print("Calificación: A")
elif puntaje >= 80 and puntaje < 90:
    print("Calificación: B")
else:
    print("Calificación: C o inferior")
```

```python
# Ejemplo 3: Sentencia if-elif-else para obtener Calificación A
# Ajustamos el puntaje para que entre en el primer bloque

puntaje = 95  # Al ser mayor o igual a 90, se cumple la primera condición

if puntaje >= 90:
    print("Calificación: A")
elif puntaje >= 80 and puntaje < 90:
    print("Calificación: B")
else:
    print("Calificación: C o inferior")
```

```python
anio = 2024
if (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
    print(f"El año {anio} es bisiesto.")
else:
    print(f"El año {anio} no es bisiesto.")
```
